creating tensors with data

```python
import torch

t_c = [0.5, 14.0, 15.0, 28.0, 11.0, 8.0, 3.0,-4.0, 6.0, 13.0, 21.0]
t_u = [35.7, 55.9, 58.2, 81.9, 56.3, 48.9, 33.9, 21.8, 48.4, 60.4, 68.4]
t_c = torch.tensor(t_c)
t_u = torch.tensor(t_u)
```

the model

```python
def model(t_u, w, b):
    return t_u*w + b
```

loss function

```python
def loss_function(t_p, t_c):
    squared_loss = (t_p-t_c)**2
    return squared_loss.mean()
```

running

```python
w = torch.ones(())
b = torch.zeros(())

t_p = model(t_u, w, b)
t_p
```

```python
loss = loss_function(t_p, t_c)
loss
```

trying to change paramenters without the derivatives, this only works with very few parameters

```python
delta = 0.1
loss_rate_of_change_w = (loss_function(model(t_u, w + delta, b), t_c) - loss_function(model(t_u, w- delta, b), t_c)) / (2.0 * delta)

learning_rate = 1e-2
w = w-learning_rate * loss_rate_of_change_w

loss_rate_of_change_b = (loss_function(model(t_u, w, b + delta), t_c) - loss_function(model(t_u, w, b- delta), t_c)) / (2.0 * delta)
b = b-learning_rate * loss_rate_of_change_b

w,b
```

```python
#derivative of loss function

def dloss_fn(t_p, t_c):
    dsq_diffs = 2 * (t_p- t_c) / t_p.size(0)  # dont know how this works yet
    return dsq_diffs

# derivatives of model wrt w and b

def dmodel_dw(t_u, w, b):
    return t_u

def dmodel_db(t_u, w, b):
    return 1.0


# gradiant function to calculate the final gradiant that will be multiplied with learning rate and subtracted from the parameters

def grad_fn(t_u, t_c, t_p, w, b):     # still need to take another look at this function but i get the general idea
    dloss_dtp = dloss_fn(t_p, t_c)
    dloss_dw = dloss_dtp * dmodel_dw(t_u, w, b)
    dloss_db = dloss_dtp * dmodel_db(t_u, w, b)
    return torch.stack([dloss_dw.sum(), dloss_db.sum()]) 
    # the book says The summation is the reverse of the broadcasting we implicitly do when applying the parameters to an entire vector of inputs in the model.
    
```

the final loop to simulate epochs and basically run the model

```python
def training_loop(n_epochs, learning_rate, params, t_u, t_c):
    for epoch in range(1, n_epochs + 1):
        w, b = params
        t_p = model(t_u, w, b)
        loss = loss_function(t_p, t_c)
        grad = grad_fn(t_u, t_c, t_p, w, b)
        params = params- learning_rate * grad
        print('Epoch %d, Loss %f' % (epoch, float(loss)))
    return params
```

running the model

```python
training_loop(n_epochs = 100, learning_rate = 1e-2, params = torch.tensor([1.0, 0.0]),t_u = t_u, t_c = t_c)
```
