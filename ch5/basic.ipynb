creating tensors with data

```python
import torch

t_c = [0.5, 14.0, 15.0, 28.0, 11.0, 8.0, 3.0,-4.0, 6.0, 13.0, 21.0]
t_u = [35.7, 55.9, 58.2, 81.9, 56.3, 48.9, 33.9, 21.8, 48.4, 60.4, 68.4]
t_c = torch.tensor(t_c)
t_u = torch.tensor(t_u)
```

the model

```python
def model(t_u, w, b):
    return t_u*w + b
```

loss function

```python
def loss_function(t_p, t_c):
    squared_loss = (t_p-t_c)**2
    return squared_loss.mean()
```

running

```python
weights = torch.ones(())
bias = torch.zeros(())

t_p = model(t_u, weights, bias)
t_p
```

```python
loss = loss_function(t_p, t_c)
loss
```
